```python
import cv2
import mediapipe as mp
import time
```

```python
print(mp.__version__)
```

```python
cap = cv2.VideoCapture(0)
mpHands = mp.solutions.hands
hands = mpHands.Hands()
```

```python
while True:
    success, img = cap.read()
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    result = hands.process(imgRGB)



    cv2.imshow("Image", img)
    cv2.waitKey(1)
```
