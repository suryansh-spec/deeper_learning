```python
from bs4 import BeautifulSoup

with open("saved_resources.html", "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

slides = soup.find_all("div", class_="slide")

for i, slide in enumerate(slides):
    title = slide.find(["h1", "h2"])
    print(f"Slide {i+1}: {title.text if title else 'No title'}")
```

```python
from pptx import Presentation

prs = Presentation()

for slide_html in slides:
    slide_layout = prs.slide_layouts[1]
    slide = prs.slides.add_slide(slide_layout)

    title = slide_html.find(["h1", "h2"])
    content = slide_html.get_text()

    slide.shapes.title.text = title.text if title else "Slide"
    slide.placeholders[1].text = content[:500]  # trim text

prs.save("output.pptx")
```
